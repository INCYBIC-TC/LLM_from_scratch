### Project Implementation without Pipeline

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv("DATASET/titanic.csv")
```

```python
df
```

```python
# drop unnecessary column 
df.drop(columns=['PassengerId','Ticket','Name','Cabin'],inplace=True)
```

```python
df.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['Survived']),
                                                df['Survived'],test_size=0.2,random_state=42)
```

```python
df.isnull().sum()
```

```python
# apply simple Imputer on age column
si_age = SimpleImputer()
si_embarked =SimpleImputer(strategy='most_frequent')

X_train_age =si_age.fit_transform(X_train[['Age']])
X_train_embarked =si_embarked.fit_transform(X_train[['Embarked']])

X_test_age =si_age.transform(X_test[['Age']])
X_test_embarked =si_embarked.transform(X_test[['Embarked']])
```

```python
# one hot encoding on Sex and embarked
one_S = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
one_E = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
X_train_S =one_S.fit_transform(X_train[['Sex']])
X_train_E =one_E.fit_transform(X_train[['Embarked']])

X_test_S =one_S.transform(X_test[['Sex']])
X_test_E =one_E.transform(X_test[['Embarked']])
```

```python
X_train_remain = X_train.drop(columns=['Sex','Age','Embarked'])
X_test_remain =X_test.drop(columns=['Sex','Age','Embarked'])
```

```python
# concatenate karo
X_train_transformed =np.concatenate((X_train_remain,X_train_age,X_train_E,X_train_S),axis=1)
X_test_transformed =np.concatenate((X_test_remain,X_test_age,X_test_E,X_test_S),axis=1)
```

```python
X_train_transformed
```

```python
model = DecisionTreeClassifier()
model.fit(X_train_transformed,y_train)
```

```python
y_pred  = model.predict(X_test_transformed)
y_pred
```

```python
from sklearn.metrics import accuracy_score
acc_score = accuracy_score(y_test,y_pred)
```

```python
acc_score
```

```python
#  save the model 
import  pickle
pickle.dump(one_S,open('models/OHE_sex.pkl','wb'))
pickle.dump(one_E,open('models/OHE_embarked.pkl','wb'))
pickle.dump(model,open('models/model_clf.pkl','wb'))
```

### Project Implementation using pipeline

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline,make_pipeline
from sklearn.feature_selection import SelectKBest,chi2
```

```python
df = pd.read_csv("DATASET/titanic.csv")
df
```

```python
df.drop(columns=['PassengerId','Ticket','Name','Cabin'],inplace=True)
```

```python
df.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['Survived']),
                                                df['Survived'],test_size=0.2,random_state=42)
```

```python
print(type(X_train))
```

```python
X_train
```

```python
y_train
```

```python
# imputation transformer 
# yaha pe you can use column name but jab transformation hoga to np array banega na ki dataframe 
# code may break in pipeline so use index
t1 =ColumnTransformer([
    ('impute_age',SimpleImputer(),[2]),
    ('impute_embarked',SimpleImputer(strategy='most_frequent'),[6])
],remainder='passthrough')
```

```python
# encoding transformer 
t2 =ColumnTransformer([
    ('one_sex_embarked',OneHotEncoder(sparse_output=False,handle_unknown='ignore'),[1,6])
],remainder='passthrough')
```

```python
df['Sex'].unique()
```

```python
df['Embarked'].unique()
```

```python
# scaling transformer
# now apne pas 8 col hai after encoding sex ke 2 column banege and embarked ke 3
# to 8 maise 2 col gaye bache 6 and ye encoding ke bad ke 5 so 11 
t3 = ColumnTransformer([
    ('scale',MinMaxScaler(),slice(0,11))
])
```

```python
#feature selection
t4 =SelectKBest(score_func=chi2,k=5)
```

```python
# train a model
t5 =DecisionTreeClassifier()
```

```python
# create a pipeline
training_pipeline =Pipeline([
    ('t1',t1),
    ('t2',t2),
    ('t3',t3),
    ('t4',t4),
    ('t5',t5)
])
```

```python
# using make_pipeline function
train_pipeline = make_pipeline(t1,t2,t3,t4,t5)
```

```python
# display pipeline 
from sklearn import set_config
set_config(display='diagram')
```

```python
# fit your pipeline
# isme agar model train nhi kar rhe to fit_transform use karo 

training_pipeline.fit(X_train,y_train)
```

```python
training_pipeline.steps# this gives an array (named_steps give an dict)
```

```python
# predict 
y_pred = training_pipeline.predict(X_test)

from sklearn.metrics import accuracy_score
acc_sc = accuracy_score(y_test,y_pred)
print(acc_sc)
```

```python
# cross validation using pipeline
from sklearn.model_selection import cross_val_score
cross_val_score(training_pipeline,X_train,y_train,scoring='accuracy').mean()
```

```python
# grid search using pipeline
params ={
    't5__max_depth':[1,2,3,4,5,None]
}
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(training_pipeline,params,cv=5,scoring='accuracy')
grid.fit(X_train,y_train)
```

```python
grid.best_score_
```

```python
grid.best_params_
```

```python
# Exporting pipeline
import pickle
pickle.dump(training_pipeline,open('models/pipe.pkl','wb'))
```
