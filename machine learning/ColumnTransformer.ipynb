# Column Transformation

```python
# now we perform encoding in traditional way 
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv("machine learning/DATASET/covid_toy.csv")
```

```python
df
```

```python
# get your dependent and independent feature 
X =df.drop(columns=['has_covid'])
y =df['has_covid']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# adding simple immputer 
si = SimpleImputer()
X_train_fever =si.fit_transform(X_train[['fever']])
X_test_fever =si.transform(X_test[['fever']])
```

```python
# apply ordinal encoding 
ordinal = OrdinalEncoder(categories=[['Mild','Strong']])
X_train_cough =ordinal.fit_transform(X_train[['cough']])
X_test_cough =ordinal.transform(X_test[['cough']])
```

```python
# apply one hot encoding
ohe = OneHotEncoder(drop ='first',sparse_output=False)
X_train_gender_city = ohe.fit_transform(X_train[['gender','city']])
X_test_gender_city = ohe.transform(X_test[['gender','city']])
```

```python
# extract the age column 
X_train_age = X_train[['age']].values
X_test_age = X_test[['age']].values
```

```python
# concatenate the all transformed column 
X_train_transformed = np.concatenate((X_train_age, X_train_fever, X_train_gender_city, X_train_cough), axis=1)
X_test_transformed = np.concatenate((X_test_age, X_test_fever, X_test_gender_city, X_test_cough), axis=1)
```

```python
X_train_transformed.shape
```

```python
from sklearn.compose import ColumnTransformer
transformer = ColumnTransformer(
    transformers=[
        ('t1',SimpleImputer(),['fever']),
        ('t2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
        ('t3',OneHotEncoder(drop='first',sparse_output=False),['gender','city'])
    ],remainder='passthrough')
```

```python
transformer.fit_transform(X_train)
transformer.transform(X_test)
```
